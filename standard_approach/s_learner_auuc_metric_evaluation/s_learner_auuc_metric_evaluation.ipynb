# **S-learner**

The S-learner (Single learner) is the most straightforward way an ML model estimates casual effects (marketing campaign or discount actually causes a customer to stay)

## **The Churn Scenario Setup**

Imagine you run a subscription software company and want to know: Does offering a $20 discount code ($X$) actually prevent customers from churning ($Y$)?

In your historical data, you have:
- Treatment ($X$): $1$ if the customer received a discount, $0$ if they did not.
- Outcome ($Y$): $1$ if the customer churned (canceled), $0$ if they stayed. Because $Y$ is binary, our machine learning model will be a classifier (like XGBoost or Logistic Regression) that predicts the probability of churning ($P(Y=1)$).
- Confounders ($Z_1, Z_2$): Variables that affect both whether a customer gets a discount and whether they churn.
    - $Z_1$: Customer tenure (months subscribed).
    - $Z_2$: App logins per week.
    
(Why are $Z_1$ and $Z_2$ confounders? Because your retention team historically gave discounts mostly to low-usage, high-risk customers, creating selection bias.)

## **Implementation**

### **Training and Testing of the Model**

```python
# Import necessary libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split,GridSearchCV, StratifiedKFold
from sklearn.metrics import  recall_score, precision_score, f1_score,roc_auc_score
```

**Generate synthetic data for demonstration**

```python
def generate_enterprise_churn_data(n=10000, seed=42):
    np.random.seed(seed)
    
    # 1. Realistic Confounders
    # Tenure: Gamma distribution (skewed: most customers are newer, fewer are multi-year veterans)
    tenure_months = (np.clip(np.random.gamma(shape=2, scale=6, size=n), 1, 60)).astype(int)
    
    # Weekly Logins: Poisson distribution
    weekly_logins = np.random.poisson(lam=4, size=n)
    
    # Support Tickets: Higher tickets usually indicate friction
    support_tickets = np.random.poisson(lam=1.5, size=n)
    
    # Monthly Spend Tier ($10, $25, $50, $100)
    monthly_spend = np.random.choice([10, 25, 50, 100], size=n, p=[0.4, 0.3, 0.2, 0.1])
    
    # 2. Confounding Mechanism (Propensity Score)
    # Retention teams historically give discounts to customers with low tenure, low logins, AND high support tickets.
    propensity_logit = -0.5 + (0.05 * tenure_months) - (0.20 * weekly_logins) + (0.40 * support_tickets) - (0.005 * monthly_spend)
    discount_prob = 1 / (1 + np.exp(-propensity_logit))
    treatment_x = np.random.binomial(1, discount_prob)
    
    # 3. Outcome Mechanism (Churn)
    # True Causal Effect: Offering a discount reduces the log-odds of churning by -0.35 (~15-20% absolute drop)
    true_causal_effect = -0.35
    
    baseline_logit = (
        1.2 
        - (0.06 * tenure_months) 
        - (0.15 * weekly_logins) 
        + (0.50 * support_tickets) 
        - (0.01 * monthly_spend)
        + (true_causal_effect * treatment_x) # The true causal impact!
    )
    churn_prob = 1 / (1 + np.exp(-baseline_logit))
    outcome_y = np.random.binomial(1, churn_prob)
    
    df = pd.DataFrame({
        'tenure_months': tenure_months,
        'weekly_logins': weekly_logins,
        'support_tickets': support_tickets,
        'monthly_spend': monthly_spend,
        'discount_given': treatment_x, # Treatment (X)
        'churned': outcome_y           # Outcome (Y)
    })
    
    return df
```

```python
df = generate_enterprise_churn_data(n=10000)
```

```python
df.head()
```

**Split the data**

```python
# train test split
X = df.drop(columns=['churned'])
y = df['churned']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)
```

```python
print(f"Dataset Split Sizes:")
print(f" - Train Set: {len(X_train)} rows")
print(f" - Test Set: {len(X_test)} rows\n")
```

**Train S-learner**

```python
# from sklearn import metrics
# metrics.get_scorer_names()
```

```python
# Define Model
# https://scikit-learn.org/1.8/modules/generated/sklearn.ensemble.RandomForestClassifier.html
rf = RandomForestClassifier(random_state=50)

# Define Cross-Validation Strategy
# This takes care of train and validaition splits for us
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=50)

# Define Hyperparameter Grid
param_grid = {
    'n_estimators': [50,150, 250],
    'max_depth': [10, 15, 20]
}

# Define Scoring Metrics
# https://scikit-learn.org/stable/auto_examples/model_selection/plot_multi_metric_evaluation.html#
# https://scikit-learn.org/stable/modules/model_evaluation.html#string-name-scorers
scoring_metrics = {
    'AUC': 'roc_auc',
    'R': 'recall',
    'P': 'precision',
    'F1': 'f1'
}
# Define Grid Search
grid = GridSearchCV(
    estimator = rf,
    param_grid = param_grid,
    scoring = scoring_metrics,
    cv = cv,
    refit = "AUC", # Refit the model based on the best AUC score
    n_jobs = -2, # Use all available cores except one
    verbose = 3
)
```

```python
grid
```

```python
print("Starting Grid Search...")
grid.fit(X_train, y_train)
print("Grid Search Completed.")
```

```python
print(f"\nBest Estimator: {grid.best_estimator_}\n")
print(f"Best Cross-Validation ROC AUC: {np.round(grid.best_score_, 2)}")
```

**Evaluate on test data**

```python
best_model = grid.best_estimator_
```

```python
# The 1 means we are taking the probability of the positive class (churn = 1)
test_prediction_proba = best_model.predict_proba(X_test)[:, 1]
test_prediction = best_model.predict(X_test)
```

```python
# Evaluations
test_auc = roc_auc_score(y_test, test_prediction_proba)
test_recall = recall_score(y_test, test_prediction)
test_precision = precision_score(y_test, test_prediction)
test_f1 = f1_score(y_test, test_prediction)
```

```python
print("Test Set Evaluation Metrics:")
print(f"ROC-AUC: {np.round(test_auc, 2)}")
print(f"Recall: {np.round(test_recall, 2)}")
print(f"Precision: {np.round(test_precision, 2)}")
print(f"F1-Score: {np.round(test_f1, 2)}")
```

### **Counterfactuals (ATE): S-Learner Approach**

So the model has been trained and tested. Now we Simulate A/B groups 

```python
# Group A: Discount Given
test_x1 = X_test.copy()
test_x1['discount_given'] = 1
prob_y1 = best_model.predict_proba(test_x1)[:, 1]
```

```python
# Group B: No Discount Given
test_x2 = X_test.copy()
test_x2['discount_given'] = 0
prob_y2 = best_model.predict_proba(test_x2)[:, 1]
```

```python
# Calculate Average Treatment Effect (ATE)
ate = np.round(np.mean(prob_y1 - prob_y2),3)
print(f"Average Treatment Effect (ATE) of Discount on Churn Probability: \n{ate}")
```

**Inspection of Prediction group A**

```python
test_x1['predicted_churn'] = best_model.predict(test_x1)
```

```python
test_x1['actual_discount'] = X_test['discount_given']
test_x1['actual_churn'] = y_test
test_x1.head()
```

**Inspection of Prediction group B**

```python
test_x2['predicted_churn'] = best_model.predict(test_x2)
```

```python
test_x2['actual_discount'] = X_test['discount_given']
test_x2['actual_churn'] = y_test
test_x2.head()
```

### **Interpretation of ATE**

Since estimated Average Treatment Effect (ATE) is **-0.058** (or roughly **-5.8%**), it means:

> **Offering the discount causes an absolute 5.8% reduction in the probability of customer churn**, on average, after properly controlling for all confounders (like tenure, app logins, support tickets, and monthly spend).

So what does it actually mean?

1. **Direction (Negative):** Because churn represents "leaving" ($1 = \text{churned}, 0 = \text{stayed}$), a **negative** causal effect is a **good thing**. It means the discount is working; it pushes the probability of canceling downward.
2. **Magnitude (5.8 Percentage Points):** It is an *absolute percentage point change*, not a relative percentage. For example:
    - If a customer's baseline predicted churn risk without a discount was **20.0%**, giving them the discount drops their predicted churn risk to **14.2%** ($20.0\% - 5.8\%$).

3. **Causally Isolated:** Because the S-learner (backed by the machine learning model and cross-validation) controlled for selection bias, this isn't just a coincidence or correlation. It proves that the retention team's discount program actively saved customers who might have otherwise left.

### **Calculating AUUC (Area Under the Uplift Curve)**

```python
# 1. Calculate individual uplift score for each user in the test set
# (No-discount churn prob) - (Discount churn prob)
# Higher score = higher expected reduction in churn from the discount
uplift_score = prob_y2 - prob_y1
```

```python
uplift_score
```

```python
# 2. Create an evalutaion dataframe
eval_df = pd.DataFrame(
    {
    'uplift_score': uplift_score,
    'treatment': X_test['discount_given'].values,
    'churned': y_test.values 
    }
)
```

```python
# 3. Sort customers from high to low predicted uplift scores
eval_df = eval_df.sort_values(by='uplift_score', ascending=False).reset_index(drop=True)    
eval_df.head()
```

```python
# 4. Calculate cumulative gains (incremental retained customers)
# Note: Since churned=1 is bad, retained=1-churned is good.
eval_df['retained'] = 1 - eval_df['churned']
eval_df.head()
# in retained column, 1 is good, 0 is bad
```

```python
# Simulate cumulative tracking
n_total = len(eval_df)
eval_df['cumulative_population'] = np.arange(1, n_total + 1) / n_total
eval_df.head()
```

```python
eval_df.tail()
```

```python
# Separate treated and control counts and cumulative positive outcomes
# (Simplified Qini / Uplift cumulative gain calculation)
treated = eval_df["treatment"] == 1
control = eval_df["treatment"] == 0

eval_df['cum_treated'] = treated.cumsum()
eval_df['cum_control'] = control.cumsum()
```

```python
eval_df.head()
```

```python
eval_df['cum_success_treated'] = (eval_df['retained'] * treated).cumsum()
eval_df['cum_success_control'] = (eval_df['retained'] * control).cumsum()
```

```python
eval_df.head()
```

```python
# Uplift Gain at each percentile
# (Normalized difference between treated success rate and control success rate)
# To avoid division by zero safely:
control_ratio = np.maximum(eval_df['cum_control'], 1)
treated_ratio = np.maximum(eval_df['cum_treated'], 1)

eval_df.head()
```

```python
eval_df['uplift_gain'] = (eval_df['cum_success_treated'] / treated_ratio) - \
                         (eval_df['cum_success_control'] / control_ratio)
```

```python
eval_df.head()
```

```python
eval_df.tail()
```

```python
# 5. Calculate AUUC (Area Under the Uplift Curve using Trapezoidal rule)
# auuc = np.trapz(eval_df['uplift_gain'], eval_df['cumulative_population'])
# AttributeError: module 'numpy' has no attribute 'trapz'
auuc = np.trapezoid(eval_df['uplift_gain'], eval_df['cumulative_population'])

print(f"Area Under the Uplift Curve (AUUC): {np.round(auuc,3)}")
```

```python
# Apply clean professional styling
sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 6))

# 1. Plot the Model's Uplift / Qini Curve using Seaborn
sns.lineplot(
    data=eval_df, 
    x='cumulative_population', 
    y='uplift_gain', 
    label=f'S-Learner Model (AUUC = {np.round(auuc, 3)})',
    linewidth=2.5,
    color='royalblue'
)

# 2. Plot the Random Baseline 
# A random baseline scales linearly from (0,0) to the final cumulative gain at 100% population
max_pop = eval_df['cumulative_population'].max()
max_gain = eval_df['uplift_gain'].iloc[-1]
plt.plot(
    [0, max_pop], 
    [0, max_gain], 
    color='gray', 
    linestyle='--', 
    linewidth=1.5, 
    label='Random Baseline'
)

# 3. Chart Layout & Labels
plt.title('Cumulative Qini / Uplift Curve', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Cumulative Proportion of Targeted Population', fontsize=12)
plt.ylabel('Cumulative Incremental Gain (Retained Users)', fontsize=12)
plt.legend(loc='upper right', frameon=True, fontsize=10)
plt.tight_layout()

plt.show()
```

Here is how to read that Qini curve through a production lens:

* **Policy Efficiency (The Front-Loaded Slope):** The steep gradient on the left-hand side proves our S-learner is successfully ranking *persuadables*. We capture the vast majority of our incremental retention lift within the top deciles before the curve flattens.
* **Operational Alignment (The 20% Budget Cutoff):** When marketing restricts the discount to the top 20% (our 80th-percentile threshold), our operating point sits squarely on that high-yield inflection zone. We maximize capital efficiency while starving "sure things" and negative-uplift "sleeping dogs" of unnecessary promo spend.
* **The AUUC Delta ($0.068$):** The robust area trapped between our model curve and the random baseline validates our causal policy. We aren't just optimizing standard classification; we're optimizing *incremental lift*, fully justifying the production rollout.

```python
# ---------------------------------------------------------
# 1. Simulate Incoming Production Data (New, Untreated Users)
# ---------------------------------------------------------
# These are real-time customers walking through your app. 
# Notice there is NO 'discount_given' column here because they haven't gotten one yet.
production_incoming_df = pd.DataFrame({
    'tenure_months': [2, 45, 3, 5, 2],
    'weekly_logins': [7, 8, 2, 4, 1],
    'support_tickets': [2, 4, 1, 11, 6],
    'monthly_spend': [25.0, 5.0, 67.0, 70.0, 50.0]
})
```

```python
# ---------------------------------------------------------
# 2. Run Counterfactual Inference using the Trained S-Learner
# ---------------------------------------------------------
# Scenario A: What if we DO give them a discount? (discount_given = 1)
prod_scenario_treated = production_incoming_df.copy()
prod_scenario_treated['discount_given'] = 1
prob_treated = best_model.predict_proba(prod_scenario_treated)[:, 1]

# Scenario B: What if we DO NOT give them a discount? (discount_given = 0)
prod_scenario_control = production_incoming_df.copy()
prod_scenario_control['discount_given'] = 0
prob_control = best_model.predict_proba(prod_scenario_control)[:, 1]
```

```python
print(prob_control)
print(prob_treated)
```

```python
# ---------------------------------------------------------
# 3. Calculate Individual Uplift Score
# ---------------------------------------------------------
# (Control churn prob) - (Treated churn prob) = Expected drop in churn risk
production_incoming_df['uplift_score'] = prob_control - prob_treated
```

```python
# ---------------------------------------------------------
# 4. Production Targeting Decision (Top Decile / Threshold)
# ---------------------------------------------------------
# Let's say our budget only allows us to give discounts to the top 20% (80th percentile) 
# of customers who actually benefit from it.
threshold = np.percentile(production_incoming_df['uplift_score'], 80)

print(threshold)
production_incoming_df['should_target_with_discount'] = (
   production_incoming_df['uplift_score'].apply(
       lambda x: "yes" if x >= threshold else "no"
   )
)
```
