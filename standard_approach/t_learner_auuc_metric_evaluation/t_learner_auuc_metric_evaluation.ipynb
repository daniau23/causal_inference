# **T-learner**

The T-learner (Two learner) solves the S-learner's biggest weakness; where tree-based models can accidentally ignore the treatment variable ($X$) during training because powerful confounders ($Z$) overshadow it. By training two completely independent models, the T-learner forces the algorithm to account for the treatment.

## **The Churn Scenario Setup**

Imagine you run a subscription software company and want to know: Does offering a $20 discount code ($X$) actually prevent customers from churning ($Y$)?

In your historical data, you have:
- Treatment ($X$): $1$ if the customer received a discount, $0$ if they did not.
- Outcome ($Y$): $1$ if the customer churned (canceled), $0$ if they stayed.- Confounders ($Z$): Variables that affect both whether a customer gets a discount and whether they churn.
    - Tenure (months subscribed)
    - Weekly app logins
    - Support tickets
    - Monthly spend tier

## **Implementation**

### **Training and Testing of the Model**

```python
# Import necessary libraries
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split,GridSearchCV, StratifiedKFold
from sklearn.metrics import  recall_score, precision_score, f1_score,roc_auc_score
```

**Generate synthetic data for demonstration**

```python
def generate_enterprise_churn_data(n=10000, seed=42):
    np.random.seed(seed)
    
    # 1. Realistic Confounders
    # Tenure: Gamma distribution (skewed: most customers are newer, fewer are multi-year veterans)
    tenure_months = (np.clip(np.random.gamma(shape=2, scale=6, size=n), 1, 60)).astype(int)
    
    # Weekly Logins: Poisson distribution
    weekly_logins = np.random.poisson(lam=4, size=n)
    
    # Support Tickets: Higher tickets usually indicate friction
    support_tickets = np.random.poisson(lam=1.5, size=n)
    
    # Monthly Spend Tier ($10, $25, $50, $100)
    monthly_spend = np.random.choice([10, 25, 50, 100], size=n, p=[0.4, 0.3, 0.2, 0.1])
    
    # 2. Confounding Mechanism (Propensity Score)
    # Retention teams historically give discounts to customers with low tenure, low logins, AND high support tickets.
    propensity_logit = -0.5 + (0.05 * tenure_months) - (0.20 * weekly_logins) + (0.40 * support_tickets) - (0.005 * monthly_spend)
    discount_prob = 1 / (1 + np.exp(-propensity_logit))
    treatment_x = np.random.binomial(1, discount_prob)
    
    # 3. Outcome Mechanism (Churn)
    # True Causal Effect: Offering a discount reduces the log-odds of churning by -0.35 (~15-20% absolute drop)
    true_causal_effect = -0.35
    
    baseline_logit = (
        1.2 
        - (0.06 * tenure_months) 
        - (0.15 * weekly_logins) 
        + (0.50 * support_tickets) 
        - (0.01 * monthly_spend)
        + (true_causal_effect * treatment_x) # The true causal impact!
    )
    churn_prob = 1 / (1 + np.exp(-baseline_logit))
    outcome_y = np.random.binomial(1, churn_prob)
    
    df = pd.DataFrame({
        'tenure_months': tenure_months,
        'weekly_logins': weekly_logins,
        'support_tickets': support_tickets,
        'monthly_spend': monthly_spend,
        'discount_given': treatment_x, # Treatment (X)
        'churned': outcome_y           # Outcome (Y)
    })
    
    return df
```

```python
df = generate_enterprise_churn_data(n=10000)
```

```python
df.head()
```

**Split the data**

```python
# train test split
X = df.drop(columns=['churned'])
y = df['churned']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)
```

```python
print(f"Dataset Split Sizes:")
print(f" - Train Set: {len(X_train)} rows")
print(f" - Test Set: {len(X_test)} rows\n")
```

**Prepare Subsets for T-Learner**

```python
# Note: In a T-learner, 'discount_given' is EXCLUDED from the feature set 
# because the treatment is split into two separate models entirely.
features = ['tenure_months', 'weekly_logins', 'support_tickets', 'monthly_spend']
```

```python
# Split the training data into two groups based on treatment
train_mask_0 = (X_train['discount_given'] == 0)
train_mask_1 = (X_train['discount_given'] == 1)

X_train_0 = X_train.loc[train_mask_0, features]
y_train_0 = y_train[train_mask_0]

X_train_1 = X_train.loc[train_mask_1, features]
y_train_1 = y_train[train_mask_1]
```

```python
print(f"Training Subsets:")
print(f" - Control Group (X=0):   {len(X_train_0)} rows")
print(f" - Treatment Group (X=1): {len(X_train_1)} rows\n")
```

**Train Model f0 (Control Group: No Discount)**

```python
rf_0 = RandomForestClassifier(random_state=50)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=50)
param_grid = {
    'n_estimators': [50, 150, 250],
    'max_depth': [10, 15, 20]
}
scoring_metrics = {'AUC': 'roc_auc', 'R': 'recall', 'P': 'precision', 'F1': 'f1'}

grid_0 = GridSearchCV(
    estimator=rf_0, 
    param_grid=param_grid, 
    scoring=scoring_metrics,
    cv=cv, refit="AUC", 
    n_jobs=-2, 
    verbose=3
)
```

```python
print("Starting Grid Search for Model f0 (Control)...")
grid_0.fit(X_train_0, y_train_0)
best_model_0 = grid_0.best_estimator_
print(f"\nBest Estimator: {best_model_0}\n")
print(f"Model f0 Best CV ROC AUC: {np.round(grid_0.best_score_, 2)}")
```

**Train Model f1 (Treatment Group: Discount)**

```python
rf_1 = RandomForestClassifier(random_state=50)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=50)
param_grid = {
    'n_estimators': [50, 150, 250],
    'max_depth': [10, 15, 20]
}
scoring_metrics = {'AUC': 'roc_auc', 'R': 'recall', 'P': 'precision', 'F1': 'f1'}

grid_1 = GridSearchCV(
    estimator=rf_1, 
    param_grid=param_grid, 
    scoring=scoring_metrics,
    cv=cv, refit="AUC", 
    n_jobs=-2, 
    verbose=3
)
```

```python
print("Starting Grid Search for Model f1 (Treatment)...")
grid_1.fit(X_train_1, y_train_1)
best_model_1 = grid_1.best_estimator_
print(f"\nBest Estimator: {best_model_1}\n")
print(f"Model f1 Best CV ROC AUC: {np.round(grid_1.best_score_, 2)}")
```

### **Counterfactuals (ATE): T-Learner Approach**

**Simulate A/B Counterfactuals on Test Set**

```python
X_test_features = X_test[features]

# World A: Predict churn probability if EVERYONE got the discount (using Model f1)
prob_y1 = best_model_1.predict_proba(X_test_features)[:, 1]

# World B: Predict churn probability if NO ONE got the discount (using Model f0)
prob_y0 = best_model_0.predict_proba(X_test_features)[:, 1]

# Calculate Average Treatment Effect (ATE)
# This is like from s-learner: prob_y1 - prob_y2;
# where prob_y1 is the predicted churn probability if treated with discount, 
# and prob_y0 is the predicted churn probability if not treated.
ate = np.round(np.mean(prob_y1 - prob_y0), 3)
print(f"Average Treatment Effect (ATE) of Discount on Churn Probability (T-Learner):\n{ate}")
```

### **Inspection of Prediction Groups**

**Inspection of Prediction group A**

```python
# Inspection of Counterfactual Group A (Trained with Treatment = 1) Gave discount
test_results_f1 = X_test.copy()
test_results_f1['predicted_churn_prob_if_treated'] = best_model_1.predict(X_test_features)
test_results_f1['actual_churn'] = y_test
print("--- Counterfactual Group A Preview ---")
test_results_f1.head()
```

**Inspection of Prediction group B**

```python
# Inspection of Counterfactual Group B (Trained with Treatment = 0) # No discount
test_results_f0 = X_test.copy()
test_results_f0['predicted_churn_prob_if_control'] = best_model_0.predict(X_test_features)
test_results_f0['actual_churn'] = y_test
print("--- Counterfactual Group B Preview ---")
test_results_f0.head()
```

### **Interpretation of ATE**

Because the T-learner isolates the treatment groups into two entirely separate models ($f_0$ and $f_1$), this **-6.9%** absolute drop is completely immune to the S-learner's risk of a tree algorithm accidentally ignoring the discount column.

Since estimated Average Treatment Effect (ATE) is **-0.069** (or roughly **-6.9%**), it means:

**What This Means for Your Business:**

1. **Stronger Causal Proof:** We can confidently tell stakeholders: *"When we give eligible high-risk customers a $20 discount, it causes an absolute **6.9 percentage point decrease** in their probability of churning."*
2. **Real-World Translation:** If a specific customer segment has a baseline predicted churn rate of **30%**, giving them the discount reduces their expected churn risk to **23.1%** ($30\% - 6.9\%$).
3. **Model Validation:** The fact that both the S-learner (-5.8%) and the T-learner (-6.9%) land in the same neighborhood tells us the causal signal is robust, stable, and not dependent on how the model was structured.

### **Calculating AUUC (Area Under the Uplift Curve)**

```python
# 1. Extract only features (dropping treatment column, exactly like T-learner setup)
X_test_features = X_test[['tenure_months', 'weekly_logins', 'support_tickets', 'monthly_spend']]
```

```python

# 2. Get counterfactual probabilities from the two independent models
prob_y0 = best_model_0.predict_proba(X_test_features)[:, 1] # No discount (Control)
prob_y1 = best_model_1.predict_proba(X_test_features)[:, 1] # Discount given (Treatment)
```

```python
# 3. Calculate individual uplift score (Control churn - Treated churn)
# Higher score = higher expected drop in churn from the discount
uplift_score = prob_y0 - prob_y1
```

```python
# 4. Create evaluation dataframe
eval_df = pd.DataFrame({
    'uplift_score': uplift_score,
    'treatment': X_test['discount_given'].values,
    'churned': y_test.values
})
```

```python
# 5. Sort customers from high to low predicted uplift scores
eval_df = eval_df.sort_values(by='uplift_score', ascending=False).reset_index(drop=True)
```

```python
# 6. Calculate cumulative gains
eval_df['retained'] = 1 - eval_df['churned']
n_total = len(eval_df)
eval_df['cumulative_population'] = np.arange(1, n_total + 1) / n_total

treated = eval_df["treatment"] == 1
control = eval_df["treatment"] == 0

eval_df['cum_treated'] = treated.cumsum()
eval_df['cum_control'] = control.cumsum()
eval_df['cum_success_treated'] = (eval_df['retained'] * treated).cumsum()
eval_df['cum_success_control'] = (eval_df['retained'] * control).cumsum()

control_ratio = np.maximum(eval_df['cum_control'], 1)
treated_ratio = np.maximum(eval_df['cum_treated'], 1)

eval_df['uplift_gain'] = (eval_df['cum_success_treated'] / treated_ratio) - \
                         (eval_df['cum_success_control'] / control_ratio)
```

```python
# 7. Calculate AUUC
auuc = np.trapezoid(eval_df['uplift_gain'], eval_df['cumulative_population'])
print(f"T-Learner Area Under the Uplift Curve (AUUC): {np.round(auuc, 3)}")
```

```python
# 8. Plot Qini Curve with Seaborn
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 6))

sns.lineplot(
    data=eval_df, 
    x='cumulative_population', 
    y='uplift_gain', 
    label=f'T-Learner Model (AUUC = {np.round(auuc, 3)})',
    linewidth=2.5,
    color='darkorange'
)

max_pop = eval_df['cumulative_population'].max()
max_gain = eval_df['uplift_gain'].iloc[-1]
plt.plot(
    [0, max_pop], [0, max_gain], 
    color='gray', linestyle='--', linewidth=1.5, label='Random Baseline'
)

plt.title('T-Learner Cumulative Qini / Uplift Curve', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Cumulative Proportion of Targeted Population', fontsize=12)
plt.ylabel('Cumulative Incremental Gain (Retained Users)', fontsize=12)
plt.legend(loc='upper right', frameon=True, fontsize=10)
plt.tight_layout()
plt.show()
```

1. Interpreting the T-Learner Qini Curve (AUUC = 0.053)
- Policy Efficiency (The Front-Loaded Slope): The sharp initial gradient on the left-hand side confirms that our dual-model architecture ($f_0$ vs. $f_1$) successfully isolates and ranks persuadables. The model front-loads the highest incremental retention gains into the top deciles before the curve flattens.
- Operational Alignment (The 20% Budget Cutoff): Restricting promotional discounts to the top 20% of users establishes an explicit operating threshold ($0.206$), ensuring that our capital is concentrated entirely on the highest-yield segment of persuadables while starving sure-things and sleeping dogs.
- The AUUC Delta ($0.053$): While slightly lower than the S-learner due to variance splitting between two independent random forest models, an AUUC of $0.053$ firmly validates our causal policy. It proves that the T-learner has captured true treatment heterogeneity, entirely free from the risk of feature-masking present in single-model setups.

## **Simulating a deployed ML model**

```python
# 1. Simulate Incoming Production Data (New, Untreated Users - NO 'discount_given' column)
production_incoming_df = pd.DataFrame({
    'tenure_months': [2, 45, 3, 5, 2],
    'weekly_logins': [7, 8, 2, 4, 1],
    'support_tickets': [2, 4, 1, 11, 6],
    'monthly_spend': [25.0, 5.0, 67.0, 70.0, 50.0]
})
```

```python
production_incoming_df.head()
```

```python
# 2. Run Counterfactual Inference using the TWO T-Learner models
# Note: We pass raw features directly without modifying a 'discount_given' column
prod_features = production_incoming_df[['tenure_months', 'weekly_logins', 'support_tickets', 'monthly_spend']]
pred_treated = best_model_1.predict(prod_features)
pred_control = best_model_0.predict(prod_features)
print(pred_control)
print(pred_treated)
```

```python
# 2. Run Counterfactual Inference using the TWO T-Learner models
# Note: We pass raw features directly without modifying a 'discount_given' column
prod_features = production_incoming_df[['tenure_months', 'weekly_logins', 'support_tickets', 'monthly_spend']]

prob_control = best_model_0.predict_proba(prod_features)[:, 1] # If NO discount
prob_treated = best_model_1.predict_proba(prod_features)[:, 1] # If discount given
```

```python
# 3. Calculate Individual Uplift Score
production_incoming_df['uplift_score'] = prob_control - prob_treated
print(prob_control)
print(prob_treated)
```

```python
# 4. Production Targeting Decision (Top 20% Budget Cutoff / 80th Percentile)
threshold = np.percentile(production_incoming_df['uplift_score'], 80)
print(f"Production Targeting Threshold (80th Percentile): {threshold}\n")

production_incoming_df['should_target_with_discount'] = production_incoming_df['uplift_score'].apply(
    lambda x: "yes" if x >= threshold else "no"
)

display_cols = ['tenure_months', 'weekly_logins', 'support_tickets', 'monthly_spend', 'uplift_score', 'should_target_with_discount']
production_incoming_df[display_cols]
```

### **Key Insights from the Production Inference Table**

1. **The Prime Persuadable (Row 2):**
* This customer exhibits a robust positive uplift score of **`0.272`**, easily clearing the 80th percentile threshold (**`0.206`**). They are correctly flagged as **`yes`** for targeting, representing a high-value candidate whose churn probability is significantly mitigated by the discount.


2. **Severe "Sleeping Dogs" (Rows 0 & 3):**
* Rows 0 and 3 show heavily negative uplift scores (**`-0.333`** and **`-0.086` respectively**), indicating that offering a discount to these profiles actually *increases* their likelihood of churning. Because the T-learner predicts independent counterfactual outcomes across separate models, it successfully protects marketing capital from alienating these sensitive segments.


3. **Budget Guardrail Filtering (Row 1):**
* Row 1 has a positive uplift score (**`0.189`**), meaning they *do* benefit from the discount. However, because our strict budget constraint caps targeting at the top 20% (threshold **`0.206`**), they are correctly categorized as **`no`**. This demonstrates how quantitative business thresholds maintain campaign ROI under limited promotional budgets.
